# Membuat Aplikasi GUI Desktop Menggunakan Python dengan Tkinter
**Tkinter** merupakan built-in library Python untuk membuat ***Graphical User Interfaces (GUIs)***. Tkinter membungkus (*wrapper*) modul *[Tcl/Tk GUI toolkit](https://www.tcl-lang.org/)*, sehingga developer Python dapat dengan mudah dan sederhana membangun desktop applications. Library ini mendukung layout management, event handling and customization, membuatnya ideal untuk rapid GUI development di Python.

Mengapa kita perlu Tkinter?
- Menyediakan 'built-in' library dan mudah digunakan untuk membuat Aplikasi GUI (Graphical User Interface) di Python.
- Menawarkan beragam 'widgets' seperti `buttons`, `labels`, `text boxes`, `menus` dan lain-lain.
- Menghilang kebutuhan external GUI frameworks karena tkinter di bundle bersama Python.
- Support event-driven programming, membuatnya cocok untuk user interfaces yang responsive

## A. Membuat Aplikasi Tkinter Pertama
Untuk membuat aplikasi Tkinter di Python, ikuti langkah dasar berikut:<br>
1. Impor modul tkinter: Impor modul `tkinter` yang diperlukan untuk membuat komponen GUI.
2. Buat window utama atau biasanya diberi nama dengan `root` : Inisialisasi `root` aplikasi menggunakan kelas `Tk()`.
3. Atur properti window : Kita dapat mengatur properti seperti judul (`title`) dan ukuran window (`geometry`).
4. Tambahkan `widget` ke `root`: Tambahkan widget seperti `button`, `label`, `entry`, dan lain-lain untuk merancang antarmuka.
5. Susun `widget`: Gunakan pengelola tata letak seperti `pack()`, `grid()` atau `place()` untuk menata widget di dalam jendela.
6. Pasang pemicu (`handler`) pada widget: Lampirkan `event handler` pada widget agar merespons interaksi pengguna (mis. `klik tombol`).

Ada dua metode utama yang perlu diingat saat membuat aplikasi GUI dengan Tkinter:

1. `Tk()`
Untuk membuat `main window` atau `root` di Tkinter, gunakan kelas `Tk()`.

    Contoh sintaks: `root = Tk(screenName=None, baseName=None, className='Tk', useTk=1)`

    Parameter:
    - `screenName` (opsional): Menentukan tampilan (umumnya untuk sistem Unix dengan beberapa layar).
    - `baseName` (opsional): Menentukan nama dasar aplikasi (default adalah nama skrip).
    - `className` (opsional): Nama kelas jendela (berguna untuk styling dan pengaturan window manager).
    - `useTk` (opsional): Boolean yang menentukan apakah subsistem Tk diinisialisasi (biasanya default `1`).

2. `mainloop()` : Metode `mainloop()` untuk menjalankan aplikasi sampai selesai (close). 
    
    Ini adalah loop tak hingga yang menjaga aplikasi tetap berjalan, menunggu dan memproses event sampai jendela ditutup.

Contoh: kode berikut menginisialisasi jendela sederhana menggunakan Tkinter. `tkinter.Tk()` membuat jendela utama, sedangkan `mainloop()` menjalankan event loop agar jendela tetap responsif.

```python
#----------------------------
import tkinter
root = tkinter.Tk()

# tambahkan widget di sini

root.mainloop()
#----------------------------
```

## C. Membuat Menu pada window utama atau window aplikasi
Cobalah buat **menu** pada **window utama atau root** dengan menambahkan `widget Menu`, `menu_name = Menu(master, option=value)`

1. Untuk pembuatan aplikasi yang baik dan modular, sebuah aplikasi sebaiknya di kemas dalam sebuah Class
ketik kode berikut: lalu jalankan (run)

````python
import tkinter as tk
from tkinter import *

# definisikan class aplikasi
# -----------------------------------------awal Class------------------------------------------
class ImageProcessingApp:
    
    #definisikan App class constructor
    def __init__(self, root):

        # inisialisasi window aplikasi
        self.root = root
        self.root.title("Aplikasi Pengolahan Citra Digital")
        self.root.geometry("800x600")

        # tambahkan menu (menuBar) ke window aplikasi(root)  
        self.menuBar = Menu(self.root)
        self.root.config(menu=self.menuBar)

        # buat item menu     
        self.menuFile = Menu(self.menuBar)
        self.menuEdit = Menu(self.menuBar)
        self.menuOpen = Menu(self.menuFile)
        self.menuSave = Menu(self.menuFile)
        self.menuExit = Menu(self.menuFile)
        
        # Menambahkan item kedalam menu:
        # - add_cascade   --> menu dengan sub_menu,
        # - add_command   --> menu tanpa sub_menu (tombol),
        # - add_separator --> pemisah
        
        # tambah item ke menuBar
        self.menuBar.add_cascade(label='File', menu=self.menuFile)
        self.menuBar.add_cascade(label='Edit', menu=self.menuEdit)

        # tambah item ke menuFile
        self.menuFile.add_command(label='Open', command=self.menuOpen_click)
        self.menuFile.add_command(label='Save', command=self.menuSave_click)
        self.menuFile.add_separator()
        self.menuFile.add_command(label='Exit', command=self.menuExit_click)

    #------akhir constructor-----------------

    #------awal fungsi untuk setiap item menu-----

    def menuOpen_click(self):
        print("Menu Open diklik")
    
    def menuSave_click(self):
        print("Menu Save diklik")
    
    def menuExit_click(self):
        print("Menu Exit diklik")
        self.root.destroy()  # menutup window aplikasi

    #------akhir fungsi untuk setiap item menu-----

# -----------------------------------akhir Class------------------------------------

#----Awal Program Utama (main)-----------
if __name__ == "__main__":
    # buat object window
    root = tk.Tk()
    
    # buat instance dari class aplikasi, dan masukkan object window nya sebagai parameter
    app = ImageProcessingApp(root)

    # jalankan window aplikasi
    root.mainloop()

#----Akhir Program Utama (main)----------- 
````
